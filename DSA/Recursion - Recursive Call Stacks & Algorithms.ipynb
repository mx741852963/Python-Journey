```python
# Fibonacci
# F(0) = 0
# F(1) = 1
# F(n) = F(n-1)+F(n-2) : n > 1
# Time : O(2^n) ,Space : O(n)
def fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)
```

```python
fibonacci(5)
```

```python
# Linked Lists
class SingleNode:
    def __init__(self, value, next=None):
        self.value = value
        self.next = next

    def __str__(self):
        return str(self.value)


Head = SingleNode(1)
A = SingleNode(2)
B = SingleNode(6)
C = SingleNode(9)
D = SingleNode(8)
Head.next = A
A.next = B
B.next = C
C.next = D
print(Head)
```

```python
# Time : O(n) ,Space : O(n)
def reverse(node):
    if node is None:
        return
    print(node)
    reverse(node.next)
    print(node)


reverse(Head)
```
