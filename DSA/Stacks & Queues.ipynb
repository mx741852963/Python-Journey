```python
# Stack -LIFO
stk = []
stk
```

```python
# Append to top of stack O(1)
stk.append(1)
stk.append(2)
stk.append(3)
stk
```

```python
# Pop from stack O(1)
x = stk.pop()
print(x)
print(stk)
```

```python
# Ask what's on the top of stack O(1)
if stk:  # not empty
    print(True)
```

```python
# Queues FIFO
from collections import deque

# Double-Ended Queue
q = deque()
q
```

```python
# Enqueue - add element to the right O(1)
q.append(1)
q.append(2)
q.append(3)
```

```python
q
```

```python
# Dequeue - Remove element from the left  O(1)
q.popleft()
```

```python
q
```

```python
# Peek from left side O(1)
q[-1]
```

```python
# Peek from right side O(1)
q[0]
```
