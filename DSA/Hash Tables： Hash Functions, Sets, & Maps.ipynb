```python
# Hashsets
s = set()
print(s)
```

```python
# Add item into set O(1)
s.add(1)
s.add(2)
s.add(3)
print(s)
```

```python
# Lookup if item in set O(1)
if 1 in s:
    print(True)
```

```python
s.remove(3)
```

```python
s
```

```python
string = 'aaofnyyuuhhddvv'
sett = set(string)  # Set construction O(s) -s is the length of the string
```

```python
sett
```

```python
# Loop over items in set O(n)
for x in s:
    print(x)
```

```python
# Hashmaps - Dictionaries
d = {'greg': 1, 'steven': 2, 'robin': 3}
d
```

```python
# Add key:value in dictionary O(1)
d["ahmad"] = 4
```

```python
d
```

```python
# Check for presence of key in dictionary O(1)
if "ahmad" in d:
    print(True)
```

```python
# Check the value corresponding to a key in the dictionary O(1)
d["ahmad"]
```

```python
# Loop over the key:value pairs of the dict O(n)
for key, value in d.items():
    print(key, value)
```
