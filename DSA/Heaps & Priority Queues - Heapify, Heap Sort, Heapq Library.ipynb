```python
# Build Min Heap (Heapify)
# Time O(n),Space O(1)
A = [5, 8, 1, 3, 4, -1, 5, -3, -7, 1, 2, -2, -3]
import heapq

heapq.heapify(A)
```

```python
# Heap Push (Insert element)
# Time O(log n),Space O(1)
heapq.heappush(A, -5)
```

```python
# Heappop (Extract min)
# Time O(log n),Space O(1)
minn = heapq.heappop(A)
A, minn
```

```python
# Heap Sort
# Time O(n log n),Space O(n)
def heap_sort(arr):
    heapq.heapify(arr)
    n = len(arr)
    new_list = [0] * n
    for i in range(n):
        minn = heapq.heappop(arr)
        new_list[i] = minn
    return new_list
```

```python
heap_sort(A)
```

```python
heap_sort([9, 8, 1, 3, 7, -7, 5, -2, 6, 1, 5, -7, 4, 6, -3, 5, -4, 5])
```

```python
# Map Heap
B = [8, 1, 3, 4, -1, 5, -3, -7, 1, 2, -2, -3]
n = len(B)
for i in range(n):
    B[i] = -B[i]
heapq.heapify(B)
B
```

```python
# Build Heap from scratch Time O(n log n) Note Heapify func Time O(n)
C = [1, 3, 9, 7, 6, 4, -1, 5, -6, -1, 1, -7, -1]
heap = []
for x in C:
    heapq.heappush(heap, x)
    print(heap, len(heap))
```

```python
# Putting tuple of items in a heap
D = [9, 8, -7, 5, -2, 6, 1, 5, -4, 5]
from collections import Counter

counter = Counter(D)
counter
```

```python
heap = []
for key, value in counter.items():
    heapq.heappush(heap, (value, key))
```

```python
# O(1) Space is possible via swaping
def sift_down(arr):
    n = len(arr)
    for i in range(n // 2 - 1, -1, -1):
        father = i
        while True:
            left = 2 * father + 1
            right = 2 * father + 2
            smallest = father
            if right < n and arr[right] < arr[smallest]:
                smallest = right
            if left < n and arr[left] < arr[smallest]:
                smallest = left

            if smallest == father:
                break
            arr[father], arr[smallest] = arr[smallest], arr[father]
            father = smallest
```

```python
t = [9, -9, 1, 0, 0, 0, 1, 5, -7, 4, 6, -3, 5, -4, 5]
sift_down(t)
t
```
