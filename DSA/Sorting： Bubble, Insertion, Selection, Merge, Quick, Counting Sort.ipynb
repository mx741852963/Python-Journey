```python
# Bubble Sort
# Time O(n^2)
# Space O(1)
def bubble_sort(arr):
    n = len(arr)
    flag = 1
    while flag:
        flag = 0
        for i in range(0, n - 1):
            if arr[i - 1] > arr[i]:
                flag = 1
                arr[i], arr[i - 1] = arr[i - 1], arr[i]


array = [3, 2, 1, 5, 6, 4, 2, 3, 1, 2, 3, 1, -4, 5, ]
bubble_sort(array)
array
```

```python
# Insertion Sort
# Time O(n^2)
# Space O(1)
def insertion_sort(arr):
    n = len(arr)
    for i in range(1, n):
        for j in range(i, 0, -1):
            print(j, end=" ")
            if arr[j - 1] > arr[j]:
                arr[j], arr[j - 1] = arr[j - 1], arr[j]
            else:
                break


B = [3, 2, 1, 5, 6, 4, 2, 3, 1, 2, 3, 1, -4, 5, ]
insertion_sort(B)
B
```

```python
# Selection Sort
# Time O(n^2)
# Space O(1)
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]


C = [3, 2, 1, 5, 6, 4, 2, 3, 1, 2, 3, 1, -4, 5]
selection_sort(C)
C
```

```python
# Merge Sort
# Time O(n log n)
# Space :O(n)
# Note can be log n but this is harder to write
D = [3, 2, 1, 5, 6, 4, 2, 3, 1, 2, 3, 1, -4, 5, ]


# D = [1,2,3,6,5,4]
def merge_sort(arr):
    n = len(arr)
    if n == 1:
        # if n == 1 or n == 0:
        return arr
    m = len(arr) // 2
    left = arr[:m]
    right = arr[m:]
    left = merge_sort(left)
    right = merge_sort(right)

    l, r = 0, 0
    l_len = len(left)
    r_len = len(right)
    sorted_arr = [0] * n

    i = 0
    while l < l_len and r < r_len:
        if left[l] < right[r]:
            sorted_arr[i] = left[l]
            l += 1
        else:
            sorted_arr[i] = right[r]
            r += 1
        i += 1
    while l < l_len:
        sorted_arr[i] = left[l]
        l += 1
        i += 1
    while r < r_len:
        sorted_arr[i] = right[r]
        r += 1
        i += 1
    return sorted_arr


merge_sort(D)
```

```python
# Quick Sort
# Time O(n log n) (Average case, technically worst case is O(n^2)
# Space :O(n)
E = [3, 2, 1, 5, 6, 4, 2, 3, 1, 2, 3, 1, -4, 5, ]


# D = [1,2,3,6,5,4]
def quick_sort(arr):
    n = len(arr)
    if n <= 1:
        return arr
    p = arr[-1]
    L = [x for x in arr[:-1] if x <= p]
    R = [x for x in arr[:-1] if x > p]
    L = quick_sort(L)
    R = quick_sort(R)
    return L + [p] + R


quick_sort(E)
```

```python
# Counting Sort
# Time O(n+k) where k is the range of data
# Note - this can be written with negative arrays , but we will stick to positive arrays,
# so k is the max value of the array
# Space: O(k)
F = [3, 2, 1, 5, 6, 4, 2, 3, 1, 2, 3, 1, 5]


def counting_sort(arr):
    maxx = max(arr)
    counts = [0] * (maxx + 1)
    for x in arr:
        counts[x] += 1
    i = 0
    for c in range(maxx + 1):
        while counts[c] > 0:
            arr[i] = c
            i += 1
            counts[c] -= 1


counting_sort(F)
F
```
