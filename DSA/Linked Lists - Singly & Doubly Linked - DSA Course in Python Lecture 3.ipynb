```python
# Single Linked Lists
class SingleNode:
    def __init__(self, value, next=None):
        self.value = value
        self.next = next

    def __str__(self):
        return str(self.value)
```

```python
Head = SingleNode(1)
A = SingleNode(2)
B = SingleNode(6)
C = SingleNode(9)
D = SingleNode(8)
Head.next = A
A.next = B
B.next = C
C.next = D
print(Head)
```

```python
# Traverse the list O(n)
current = Head
while current:
    print(current.value)
    current = current.next
```

```python
# Display the list O(n)
elements = []


def display(Head):
    current = Head
    while current:
        elements.append(str(current.value))
        current = current.next
    print("->".join(elements))


display(Head)
```

```python
# Search for Node Value O(N)
def search(head, target):
    curr = head
    while curr:
        if curr.value == target:
            return True
        curr = curr.next
    return False
```

```python
search(Head, 5)
```

```python
search(Head, 6)
```

```python
search(Head, 7)
```

```python
# Doubly Linked Lists
class DoublyNode:
    def __init__(self, value, next=None, prev=None):
        self.value = value
        self.next = next
        self.prev = prev

    def __str__(self):
        return str(self.value)
```

```python
head = tail = DoublyNode(1)
print(head)
```

```python
# Display the list O(n)
def display(Head):
    current = Head
    elements = []
    while current:
        elements.append(str(current.value))
        current = current.next
    print(" <-> ".join(elements))


display(head)
```

```python
# Insert  to beginning O(1)
def insert_into_beginning(head, tail, value):
    new_node = DoublyNode(value, next=head)
    head.prev = new_node
    return new_node, tail


```

```python
head, tail = insert_into_beginning(head, tail, 5)
```

```python
display(head)
```

```python
head, tail = insert_into_beginning(head, tail, 7)
```

```python
display(head)
```

```python
# Insert to end O(1)
def insert_into_end(head, tail, value):
    new_node = DoublyNode(value, prev=tail)
    tail.next = new_node
    return head, new_node
```

```python
head, tail = insert_into_end(head, tail, 11)
```

```python
display(head)
```

```python
head, tail = insert_into_end(head, tail, 15)
```
