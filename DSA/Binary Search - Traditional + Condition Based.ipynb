```python
A = [-9, -6, -1, 0, 1, 2, 3, 4, 7, 9]
# Naive O(n) searching
if 5 in A:
    print(True)
else:
    print(False)
```

```python
# Traditional Binary Search Looking up if number in the array
# Time O(log n)
# Space O(1)
def binary_search(arr, target):
    l = 0
    n = len(arr)
    r = n - 1
    while l <= r:
        mid = l + ((r - l) // 2)
        if arr[mid] == target:
            return True
        elif target < arr[mid]:
            r = mid - 1
        else:
            l = mid + 1
    return False

```

```python
binary_search(A, -9)
```

```python
# Condition Base
# Time O(log n)
# Space O(1)
def binary_search_condition(arr):
    l = 0
    n = len(arr)
    r = n - 1
    while l < r:
        mid = l + ((r - l) // 2)
        if arr[mid]:
            r = mid
        else:
            l = mid + 1
    return l
```

```python
B = [False, False, False, True, True]
binary_search_condition(B)
```
